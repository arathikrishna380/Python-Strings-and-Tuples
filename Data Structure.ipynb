```python
string1 = "Hello "
string2 = input("Enter your name: ")
result_string = string1 + string2
print(result_string)

string3 = " welocme to Python programming"
final_string = result_string + string3
print(final_string)
```

```python
print("a. First character:", final_string[0])
print("b. Last character:", final_string[-1])
print("c. First 5 characters:", final_string[:5])
print("d. Last 11 characters:", final_string[-11:])
print("e. Reversed string:", final_string[::-1])
start_idx = final_string.find("Python")
print("f. Extracted word:", final_string[start_idx:start_idx+6])
```

```python
strM = "Python beginner tutorial"

print("a. Uppercase:", strM.upper())
print("b. Lowercase:", strM.lower())
print("c. Capitalized:", strM.capitalize())
print("d. Count of 't':", strM.count('t'))
print("e. Replaced string:", strM.replace("Python", "Machine Learning"))
```

```python
tuple1 = (10, 20, 30)
tuple2 = (40, 50, 60)

t_combine = tuple1 + tuple2
print("a. Concatenated tuple (t_combine):", t_combine)

t_repeat = t_combine * 3
print("b. Repeated 3 times:", t_repeat)
print("c. 3rd element:", t_combine[2])
print("d. First three elements:", t_combine[:3])
print("e. Last three elements:", t_combine[-3:])
```
